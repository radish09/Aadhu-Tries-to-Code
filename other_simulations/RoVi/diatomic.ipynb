### Angular momenta in diatomics
The angular momenta that can occur in a diatomic molecule $AB$ include:
* Spin of nucleus $A$ or $B$ denoted $\vec{\textbf{I}}_A$ or $\vec{\textbf{I}}_B$.
* Orbital angular momentum of the *i*-th electron $\vec{\textbf{l}}_i$.
* Spin of the *i*-th electron $\vec{\textbf{s}}_i$.
* Rotation of the nuclear framework, $\vec{\textbf{R}}$.

| Name or description | Symbol | Formula |
| --- | --- | --- |
| total nuclear spin | $\vec{\textbf{I}}$ | $\vec{\textbf{I}} = \vec{\textbf{I}}_A + \vec{\textbf{I}}_B$ |
| total angular momentum of the *i*-th electron | $\vec{\textbf{j}}_i$ | $\vec{\textbf{j}}_i = \vec{\textbf{l}}_i + \vec{\textbf{s}}_i$ |
| total electronic orbital angular momentum | $\vec{\textbf{L}}$ | $\vec{\textbf{L}} = \sum_i \vec{\textbf{l}}_i$|
| total electron spin | $\vec{\textbf{S}}$ | $\vec{\textbf{S}} = \sum_i \vec{\textbf{s}}_i$|
| total angular momentum | $\vec{\textbf{F}}$ | $\vec{\textbf{F}} = \vec{\textbf{R}} + \vec{\textbf{L}} + \vec{\textbf{S}} + \vec{\textbf{I}}$ |
| total angular momentum exclusive of nuclear spin | $\vec{\textbf{J}}$ | $\vec{\textbf{J}} = \vec{\textbf{R}} + \vec{\textbf{L}} + \vec{\textbf{S}}$ |
| total angular momentum exclusive of nuclear and electron spin | $\vec{\textbf{N}}$ | $\vec{\textbf{N}} = \vec{\textbf{J}} - \vec{\textbf{S}} = \vec{\textbf{R}} + \vec{\textbf{L}}$ |

Angular momentum basis functions $|A \alpha M_A\rangle$ can be simultaneous eigenfunctions of three types of operators: 
* magnitude $\textbf{A}^2$  $\Leftrightarrow$ basis function label $A$
* projection of $\vec{\textbf{A}}$ onto internuclear axis, $\textbf{A}_z$ (molecule fixed) $\Leftrightarrow$ basis function label $\alpha$
* projection of $\vec{\textbf{A}}$ onto laboratory quantization axis, $\textbf{A}_Z$ (lab fixed) $\Leftrightarrow$ basis function label $M_A$

| Molecular axis projection of angular momentum | Quantum number |
| --- | --- |
| $\textbf{l}_{iz}$ | $\lambda_i = 0, 1, 2, ... $ $(\sigma, \pi, \delta, ...)$ |
| $\textbf{s}_{iz}$ | $\sigma_i = \pm \frac{1}{2}$|
| $\textbf{R}_{z}$ | zero |
| $\textbf{j}_{iz}$ | $\omega_i = \lambda_i + \sigma_i $|
| $\textbf{L}_z$ | $\Lambda = 0, 1, 2, ...$ $(\Sigma, \Pi, \Delta, ...)$ |
| $\textbf{S}_z$ | $\Sigma$ |
| $\textbf{F}_z$ | no standard notation |
| $\textbf{J}_z$ | $\Omega = \Lambda + \Sigma$ |
| $\textbf{N}_z$ | $\Lambda$ |


#### Hund's cases
| Hund's case | $\textbf{H}^\textrm{el}$ | $\textbf{H}^\textrm{SO}$ | $\textbf{H}^\textrm{ROT}$ | Basis set |
| --- | --- | --- | --- | --- |
| (a) | strong | intermediate | weak | $ \|nJS\Omega\Lambda\Sigma \rangle $|
| (b) | strong | weak | intermediate |$ \|nJSN\Lambda(S_R) \rangle $ |
| (c) | intermediate | strong | weak | $ \| n J [ J_a ] \Omega \rangle $ |
| (d) | intermediate | weak | strong | $ \| n J S N (S_R) J^+ N^+ S^+ \Lambda^+ l(l_R, s_R) \rangle$ |
| (e) | weak | intermediate | strong | $\|n J J^+ [ J^+_a \textrm{ or } S^+ \textrm{ and } \Lambda^+ ] \Omega^+ l j (j_R) \rangle $|
| (e') | weak | strong | intermediate | $\|n J J^+ [ J^+_a \textrm{ or } S^+ \textrm{ and } \Lambda^+ ] \Omega^+ l j (j_R) \rangle $ |

Quantum numbers in parentheses are good only in the high rotation limit. Quantum numbers in brackets are often not well defined. We exclude laboratory projection quantum number $M_J$ since we will not be considering laboratory-fixed electromagnetic fields.

Quantum numbers describing projection of an angular momentum onto the molecular frame rotation axis $\vec{\textbf{R}}$ are enclosed in parentheses because a basis set transformation must be performed in order to be able to make explicit use of them.

Cases (d) and (e) are for one Rydberg electron weakly coupled to a many-electron molecular ion-core. **It doesn't seem that PGOPHER has this -- maybe a possible project to do in the far future.**

#### Basis sets
Choice of basis set implies a partitioning of the Hamiltonian $\textbf{H} = \textbf{H}^\textrm{el} + \textbf{H}^\textrm{SO} + \textbf{T}^N(R) + \textbf{H}^\textrm{ROT}$ into two parts, a part $\textbf{H}^{(0)}$ that is fully diagonal in the selected basis set and a residual part $\textbf{H}^{(1)}$. If our basis is complete the choice doesn't matter, but often one basis is more convenient than another for a given problem.

The rotational Hamiltonian $\textbf{H}^\textrm{ROT} = B(R) \textbf{R}^2$ where $B(R)$ is the rotational "constant" operator and $\textbf{R}$ is the nuclear rotational angular momentum, can be expressed in a form appropriate for each of the Hund's case (a) - (e) basis sets.

**Case (a)**
$$\textbf{H}^\textrm{ROT} = B(R) (\textbf{J} - \textbf{L} - \textbf{S})^2 = B(R) [\textbf{J}^2 - \textbf{J}^2_z + \textbf{S}^2 - \textbf{S}^2_z + (\textbf{L}^2 - \textbf{L}^2_z) - \newline
\{(\textbf{J}^+ \textbf{L}^- + \textbf{J}^- \textbf{L}^+) + (\textbf{J}^+ \textbf{S}^- + \textbf{J}^- \textbf{S}^+) - (\textbf{L}^+ \textbf{S}^- + \textbf{L}^- \textbf{S}^+)\}
]$$

We replace $(\textbf{L}^2 - \textbf{L}^2_z)$ with $\textbf{L}^2_\perp$. The unevaluated expectation value of $B(R) \textbf{L}^2_\perp$ is implicitly included in the electronic energy. The zero-order rotational energy is $E^\textrm{ROT}_{(a)}  = B(R) [J(J+1) - \Omega^2 + S(S+1) - \Sigma^2]$. The three pairs of terms in $\{ \}$ give rise to off-diagonal matrix elements. These can be grouped into three sets:
* $\Delta \Omega = \Delta \lambda  = \pm 1$
* $\Delta \Omega = \Delta \Sigma  = \pm 1 $
* $\Delta \Lambda  = - \Delta \Sigma = \pm 1$

The $(\textbf{J}^+ \textbf{S}^- + \textbf{J}^- \textbf{S}^+)$ term operators within a case (a) $^{2S+1} \Lambda $ multiplet state. The other two terms are responsible for off-diagonal matrix elements between case (a) $^{2S+1} \Lambda _\Omega$ and $^{2S+1} \Lambda \pm 1$ states.

*The other cases are described in Lefebvre-Brion and Field, Chapter 3.2. I am only going to use the case (a) basis here and so will not describe these any further.*
